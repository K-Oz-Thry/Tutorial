```python
import holoviews as hv
hv.extension('bokeh')
import panel as pn
from holoviews import opts
from bokeh.plotting import figure, show, output_notebook
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from bokeh.models import ColumnDataSource, HoverTool,  LabelSet, Select, CustomJS
from bokeh.layouts import row, column
from bokeh.palettes import Category10
import plotly.express as px
import ipywidgets as widgets
output_notebook()
pn.extension()
penguins = sns.load_dataset("penguins").dropna()
mpg = sns.load_dataset("mpg").dropna()
```

### Now that we have learned from the previous libraries, let's try to make **interactive** multivariable graphics. Let's start this new journey with a favorite library of mine Bokeh.
### With Bokeh we can make the same types of graphics as we did in Seaborn or Matplotlib. The catch? these are interactive, so you could zoom in or out, make cursor movements that let you see values that are difficult to get exact with the naked eye, and more

### Let's start simple

```python
# Create a data source from the penguins DataFrame
source = ColumnDataSource(penguins)

# Create the interactive figure
p = figure(
    title="Penguins: Interactive Scatter with Labels", 
    tools="pan,wheel_zoom,box_zoom,reset"
)

# Add scatter markers
p.circle(
    "bill_length_mm", "bill_depth_mm", 
    source=source, 
    size=10, 
    color="navy", 
    alpha=0.6
)

# Add hover tool
hover = HoverTool(tooltips=[
    ("Species", "@species"),
    ("Sex", "@sex"),
    ("Body Mass", "@body_mass_g g")
])
p.add_tools(hover)

# Set axis labels
p.xaxis.axis_label = "Bill Length (mm)"
p.yaxis.axis_label = "Bill Depth (mm)"


# Show the plot
show(p)
```

### Now the graph seems great, you can move around, zoom into the values, check the stats. There's only one small issue; it all looks like its the same, even though its not. 

### With previous libraries we were able to make differences in data visualizations by using different colors, or shapes. This is something we can also do in Bokeh, but I would rather show other ways in which we can classify the data. 

### Dropdowns:
- We can take into inspiration what we did in the first notebook and create three different graphics from the previous graphic, divided by island.
- This divided graphics can be accessed by a dropdown button near the graphic.

```python
penguins = sns.load_dataset("penguins").dropna()

# Define value dimensions
vdims = [('bill_length_mm', 'Bill Length (mm)'), ('bill_depth_mm', 'Bill Depth (mm)')]

# Create HoloViews Dataset with key dimensions: island and species
ds = hv.Dataset(penguins, ['island', 'species'], vdims)

# Convert to scatter plot: x = bill length, y = bill depth, grouped by species
scatter = ds.to(hv.Scatter, 'bill_length_mm', 'bill_depth_mm', groupby='species')

# Style the scatter plot
scatter.opts(
    width=800, height=500, size=8, alpha=0.6, tools=['hover'],
    xlabel="Bill Length (mm)", ylabel="Bill Depth (mm)",
    title="Bill Length vs. Bill Depth by Species",
    color='navy', cmap='Category10'
)
```

### As you can tell the code does get a but more complex as you try to add features, but that's the trade off for get vizualizations

### And also do datamaps

### We can use a different dataset (mpg) to check plotly's ability of making datamaps 

```python
import pandas as pd
import seaborn as sns
import plotly.express as px

# Load and clean data
mpg = sns.load_dataset("mpg").dropna()

# Map 'origin' to ISO country codes
origin_to_iso = {
    "usa": "USA",
    "japan": "JPN",
    "europe": "DEU"  # Using Germany as a proxy for Europe
}

mpg["iso_code"] = mpg["origin"].map(origin_to_iso)

# Group by country
avg_mpg_by_country = mpg.groupby("iso_code").agg(avg_mpg=("mpg", "mean")).reset_index()

# Choropleth map
fig = px.choropleth(
    avg_mpg_by_country,
    locations="iso_code",
    color="avg_mpg",
    hover_name="iso_code",
    color_continuous_scale="Viridis",
    title="Average MPG by Country of Origin",
    template="plotly_white"
)
fig.update_layout(
    width=1000,  # Adjust width as needed
    height=800,  # Adjust height as needed
)
fig.show()
```

## 3D scatter plot:

```python
fig_penguins = px.box(penguins, x="species", y="body_mass_g", color="sex")
fig_penguins.show()
```

## Extra: 
- If you want to you can layout multiple graphics in the same cell

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(rows=1, cols=2, subplot_titles=("MPG Scatter", "Penguins Box"))

# Add traces
fig.add_trace(
    go.Scatter(x=mpg["horsepower"], y=mpg["mpg"], mode='markers', marker_color='blue', name="MPG"),
    row=1, col=1
)

fig.add_trace(
    go.Box(y=penguins[penguins["sex"]=="Male"]["body_mass_g"], name="Male"),
    row=1, col=2
)
fig.add_trace(
    go.Box(y=penguins[penguins["sex"]=="Female"]["body_mass_g"], name="Female"),
    row=1, col=2
)

fig.update_layout(height=400, width=900, title_text="Plotly Subplots")
fig.show()
```
